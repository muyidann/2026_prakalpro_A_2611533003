## 2.1 Dasar Teori
### 1. Python

Python merupakan bahasa pemrograman yang memiliki sintaks yang relatif sederhana dan dapat digunakan untuk membuat berbagai jenis program. Dalam praktikum ini, Python digunakan untuk menerapkan konsep perulangan, percabangan, operator, input, dan pencetakan pola.

### 2. Variabel

Variabel digunakan untuk menyimpan nilai yang dapat digunakan selama program berjalan. Pada program praktikum digunakan beberapa variabel seperti batas_3003, ulang_3003, i_3003, j_3003, tinggi_3003, dan n_3003.

Nilai dalam variabel dapat digunakan dalam perhitungan, menentukan jumlah perulangan, maupun menentukan posisi karakter dalam pola.

### 3. Input dan Konversi Tipe Data

Fungsi input() digunakan untuk menerima data dari pengguna. Secara default, data yang diterima oleh input() berupa teks. Oleh karena itu, fungsi int() digunakan untuk mengubah input menjadi bilangan bulat agar dapat digunakan dalam operasi perhitungan dan perulangan.

Contohnya digunakan pada program untuk memasukkan nilai batas atau tinggi pola.

### 4. Perulangan for

Perulangan for digunakan untuk menjalankan suatu perintah secara berulang berdasarkan suatu urutan. Pada praktikum ini, for digunakan untuk mencetak angka, karakter, menghitung nilai, dan membentuk pola.

Perulangan dapat menggunakan range() untuk menentukan nilai awal, batas akhir, dan langkah perubahan nilai.

### 5. Fungsi range()

Fungsi range() digunakan untuk menghasilkan urutan bilangan yang dapat digunakan oleh perulangan for.

Bentuk penggunaannya dapat berupa:

range(batas)

atau

range(awal, batas)

serta dapat menggunakan langkah:

range(awal, batas, langkah)

Pada program praktikum terdapat penggunaan range() yang dimulai dari 0 maupun 1. Terdapat juga penggunaan langkah negatif seperti range(n_3003, 0, -1) untuk melakukan perulangan secara menurun.

### 6. Nested for

Nested for merupakan perulangan for yang berada di dalam perulangan for lainnya. Perulangan bersarang dapat digunakan untuk mengolah data berdasarkan baris dan kolom.

Dalam praktikum, nested for digunakan untuk membuat pola bintang, pola titik, deret angka, segitiga, dan pola jam pasir. Perulangan bagian luar biasanya digunakan untuk menentukan baris, sedangkan perulangan bagian dalam digunakan untuk menentukan isi setiap baris.

### 7. Percabangan if, elif, dan else

Percabangan digunakan untuk menjalankan perintah berdasarkan kondisi tertentu. if digunakan untuk memeriksa kondisi pertama, elif digunakan untuk memeriksa kondisi tambahan, sedangkan else digunakan ketika kondisi sebelumnya tidak terpenuhi.

Pada program pola jam pasir, percabangan digunakan untuk menentukan karakter yang dicetak berdasarkan posisi baris dan kolom.

### 8. Operator Aritmatika

Operator aritmatika digunakan untuk melakukan operasi matematika. Operator yang digunakan dalam program praktikum antara lain:

+ untuk penjumlahan.
- untuk pengurangan.
* untuk perkalian.
% untuk mendapatkan sisa pembagian.

Operator tersebut digunakan untuk menghitung jumlah, menentukan posisi karakter, menghitung lebar pola, dan mengatur perubahan nilai dalam perulangan.

### 9. Operator Perbandingan

Operator perbandingan digunakan untuk membandingkan dua nilai dan menghasilkan kondisi benar atau salah. Operator yang digunakan dalam program antara lain:

== untuk memeriksa kesamaan nilai.
!= untuk memeriksa ketidaksamaan nilai.
> untuk memeriksa nilai lebih besar.
< untuk memeriksa nilai lebih kecil.
<= untuk memeriksa nilai lebih kecil atau sama dengan.

Operator perbandingan digunakan dalam percabangan untuk menentukan karakter atau proses yang harus dijalankan.

### 10. Operator Logika or dan and

Operator or digunakan ketika salah satu dari beberapa kondisi sudah cukup untuk memenuhi suatu kondisi. Operator and digunakan ketika beberapa kondisi harus terpenuhi secara bersamaan.

Dalam pola jam pasir, operator or digunakan untuk memeriksa posisi tertentu seperti baris pertama atau baris terakhir, sedangkan and digunakan untuk menentukan posisi karakter di bagian dalam pola.

### 11. Parameter end pada print()

Secara default, print() akan berpindah ke baris berikutnya setelah mencetak suatu nilai. Parameter end digunakan untuk mengubah perilaku tersebut.

Penggunaan end="" membuat hasil berikutnya tetap dicetak pada baris yang sama. Sementara itu, end=" " membuat hasil tetap pada baris yang sama dengan tambahan satu spasi.

Konsep ini digunakan untuk menyusun karakter dan angka agar membentuk pola tertentu.

### 12. Perulangan dan Akumulasi

Perulangan dapat digunakan untuk melakukan proses perhitungan berulang. Salah satunya adalah proses penjumlahan menggunakan variabel yang nilainya terus diperbarui pada setiap perulangan.

Contohnya, suatu variabel dapat dimulai dari nilai 0 kemudian ditambahkan dengan nilai i pada setiap perulangan. Dengan cara tersebut, jumlah akhir dapat diperoleh setelah seluruh perulangan selesai.

### 13. Pencetakan Pola

Pencetakan pola merupakan penerapan nested for untuk menghasilkan susunan karakter atau angka dalam bentuk tertentu. Pola dapat dibentuk dengan mengatur jumlah karakter, jumlah spasi, posisi angka, serta perpindahan baris.

Pada praktikum ini, pola yang dibuat meliputi pola titik dan angka, pola persegi menggunakan karakter *, pola hasil penjumlahan i + j, pola jam pasir, serta pola segitiga.

### 14. Pola Jam Pasir Kristal Palindromik

Pola jam pasir kristal palindromik menggunakan kombinasi perulangan, percabangan, perhitungan, dan pencetakan karakter. Program membentuk bagian atas, poros tengah, dan bagian bawah secara teratur.

Deret angka dibuat secara menurun dan kemudian menaik pada setiap baris. Simbol <*> digunakan sebagai poros kristal, sedangkan karakter #, =, dan | digunakan untuk membentuk bingkai pola.

Pengaturan jumlah spasi pada sisi kiri dan kanan digunakan untuk menjaga keseimbangan serta bentuk pola agar tetap simetris.

### 2.2.7 Program "nested_for4_2611533003"

```python
tinggi_3003 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_3003 % 2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_3003 = tinggi_3003
    c_3003 = a_3003
    lebar_3003 = (2 * tinggi_3003) - 2

    for i_3003 in range(1, tinggi_3003 + 1):
        b_3003 = c_3003 + 1

        for j_3003 in range(1, lebar_3003 + 1):

            #Baris atas dan bawah
            if i_3003 == 1 or i_3003 == tinggi_3003:
                 if j_3003 == 1 or j_3003 == lebar_3003:
                    print("#", end="")
                 else:
                    print("=", end="")
            #Baris isi
            else:
                if j_3003 == 1 or j_3003 == lebar_3003:
                    print("|", end="")
                else:
                    if j_3003 == c_3003:
                        print("<", end="")
                    elif j_3003 == b_3003:
                        print(">", end="")
                    elif j_3003 == (lebar_3003 - c_3003):
                        print("<", end="")
                    elif j_3003 == (lebar_3003 - c_3003 + 1):
                        print(">", end="")
                    elif j_3003 > b_3003 and j_3003 < (lebar_3003 - c_3003):
                        print(".", end="")
                    else:
                        print(" ", end="")
        print()

        # Logika asli Java
        a_3003 -= 2

        if a_3003 <= 0:
            c_3003 = (-a_3003) + 2
        else:
            c_3003 = a_3003
```

**Penjelasan :**
1. Baris 1 berfungsi untuk meminta pengguna memasukkan tinggi pola. Fungsi `int()` mengubah input menjadi bilangan bulat dan menyimpannya dalam variabel `tinggi_3003`.
2. Baris 3 berfungsi untuk memeriksa apakah nilai `tinggi_3003` merupakan bilangan genap. Operator `%` digunakan untuk mencari sisa pembagian, sedangkan `!= 0` berarti nilainya tidak sama dengan 0.
3. Baris 4 berfungsi untuk menampilkan pesan bahwa tinggi pola harus berupa bilangan genap jika kondisi pada baris 3 terpenuhi.
4. Baris 5 berfungsi sebagai `else`, yaitu menjalankan program apabila nilai `tinggi_3003` merupakan bilangan genap.
5. Baris 6 berfungsi untuk menyimpan nilai `tinggi_3003` ke dalam variabel `a_3003` sebagai nilai awal perhitungan pola.
6. Baris 7 berfungsi untuk menyimpan nilai `a_3003` ke dalam variabel `c_3003` yang digunakan untuk menentukan posisi karakter dalam pola.

```python
print("=== PROGRAM JAM PASIR KRISTAL PALINDROMIK (PEKAN 5) ===")

n_3003 = int(input("Masukkan ukuran skala jam pasir (N): "))


# BORDER ATAS
print("#", end="")

for jumlah_3003 in range(1):
    for karakter_3003 in range(4 * n_3003 + 5):
        print("=", end="")

print("#")


# FASE 1: JAM PASIR ATAS
for baris_3003 in range(n_3003, 0, -1):

    print("|", end="")
    print(" ", end="")

    # Spasi penyeimbang kiri
    for jumlah_3003 in range(1):
        for spasi_3003 in range(2 * (n_3003 - baris_3003)):
            print(" ", end="")

    # Deret angka mundur
    for jumlah_3003 in range(1):
        for angka_3003 in range(baris_3003, 0, -1):
            print(angka_3003, end=" ")

    # Poros kristal
    print("<*>", end="")

    # Deret angka maju
    for jumlah_3003 in range(1):
        for angka_3003 in range(1, baris_3003 + 1):
            print(" ", end="")
            print(angka_3003, end="")

    # Spasi penyeimbang kanan
    for jumlah_3003 in range(1):
        for spasi_3003 in range(2 * (n_3003 - baris_3003)):
            print(" ", end="")

    print(" ", end="")
    print("|")


# FASE 2: POROS TITIK PUSAT
print("|", end="")

for jumlah_3003 in range(1):
    for spasi_3003 in range(2 * n_3003 + 1):
        print(" ", end="")

print("<*>", end="")

for jumlah_3003 in range(1):
    for spasi_3003 in range(2 * n_3003 + 1):
        print(" ", end="")

print("|")


# FASE 3: JAM PASIR BAWAH
for baris_3003 in range(1, n_3003 + 1):

    print("|", end="")
    print(" ", end="")

    # Spasi penyeimbang kiri
    for jumlah_3003 in range(1):
        for spasi_3003 in range(2 * (n_3003 - baris_3003)):
            print(" ", end="")

    # Deret angka mundur
    for jumlah_3003 in range(1):
        for angka_3003 in range(baris_3003, 0, -1):
            print(angka_3003, end=" ")

    # Poros kristal
    print("<*>", end="")

    # Deret angka maju
    for jumlah_3003 in range(1):
        for angka_3003 in range(1, baris_3003 + 1):
            print(" ", end="")
            print(angka_3003, end="")

    # Spasi penyeimbang kanan
    for jumlah_3003 in range(1):
        for spasi_3003 in range(2 * (n_3003 - baris_3003)):
            print(" ", end="")

    print(" ", end="")
    print("|")


# BORDER BAWAH
print("#", end="")

for jumlah_3003 in range(1):
    for karakter_3003 in range(4 * n_3003 + 5):
        print("=", end="")

print("#")
```
