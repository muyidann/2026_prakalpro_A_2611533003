## 2.1 Landasan Teori

**1. Pengenalan Bahasa Pemrograman Python**

​Python merupakan bahasa pemrograman tingkat tinggi (_high-level language_) yang bersifat _interpreted_, _object-oriented_, dan _dynamically typed_. Diciptakan oleh Guido van Rossum pada tahun 1991, Python dirancang dengan penekanan pada keterbacaan kode (_code readability_) melalui penggunaan indentasi yang bersih. Sintaksisnya yang intuitif membuat Python banyak digunakan dalam berbagai bidang, mulai dari pengembangan perangkat lunak, otomatisasi sistem, hingga analisis data dan kecerdasan buatan.

​**2. Konsep Variabel dalam Python**

​Variabel merupakan nama atau label berpenunjuk (_identifier_) yang terikat pada suatu lokasi di memori komputer untuk menyimpan nilai tertentu. Dalam Python, variabel tidak perlu dideklarasikan secara eksplisit beserta tipe datanya sebelum digunakan; variabel dibuat secara otomatis saat suatu nilai dialokasikan padanya menggunakan operator penugasan (`=`). Penulisan nama variabel di Python harus mengikuti aturan tertentu, seperti bersifat case-sensitive, tidak boleh diawali oleh angka, dan tidak boleh menggunakan kata kunci (_keywords_) bawaan sistem.

**​3. Tipe Data Dasar (Primitive Data Types)**

​Tipe data mengklasifikasikan jenis nilai yang dimiliki oleh suatu variabel sehingga sistem mengetahui cara mengolah dan menyimpan data tersebut di memori. Python memiliki beberapa tipe data dasar yang sering digunakan:

> * ​Integer (`int`): Tipe data berupa bilangan bulat, baik positif maupun negatif, tanpa komponen desimal (contoh: `-5`, `0`, `100`).

> * ​Floating-Point (`float`): Tipe data berupa angka yang memiliki titik desimal atau pecahan (contoh: `3.14`, `-0.01`).

> * ​String (`str`): Tipe data berupa urutan karakter/teks yang diapit oleh tanda petik tunggal (`'...'`) atau ganda (`"..."`) (contoh: `"Praktikum Python"`).

> * ​Boolean (`bool`): Tipe data logika yang hanya memiliki dua nilai kebenaran, yaitu `True` (benar) dan `False` (salah).

**​4. Dynamic Typing dan Type Casting**

​Python menganut sistem _dynamic typing_, yang berarti tipe data dari suatu variabel ditentukan secara otomatis pada saat program berjalan (_runtime_) berdasarkan nilai yang dimasukkan, bukan saat variabel dideklarasikan. Selain itu, Python mendukung konversi tipe data atau _type casting_, yaitu proses mengubah nilai dari satu tipe data ke tipe data lainnya secara eksplisit (seperti menggunakan fungsi `int()`, `float()`, atau `str()`) maupun implisit oleh interpreter. Pemahaman terhadap type casting sangat penting untuk menghindari kesalahan tipe data (_type error_) saat melakukan operasi aritmetika atau manipulasi string.# BAB II

### 2.2.3 Program "Konstanta_2611533003"

```python
from typing import Final
PI: Final = 3.14
print(" %f" % (PI))
jari_3003 = float(input("Masukkan jari-jari lingkaran: "))
luas_3003 = PI * jari_3003 * jari_3003
print("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_3003, luas_3003))
```
